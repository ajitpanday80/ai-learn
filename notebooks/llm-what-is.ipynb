## 1. Tokenization — how LLMs read text
LLMs don't read characters or words — they read *tokens*. Let's see what that looks like.

```python
!pip install transformers -q
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained('gpt2')

texts = [
    'The capital of France is Paris.',
    'LLMs learn by predicting the next token.',
    'Hello world!',
    'ChatGPT',
    'antidisestablishmentarianism',
]

for text in texts:
    tokens = tokenizer.encode(text)
    decoded = [tokenizer.decode([t]) for t in tokens]
    print(f'Text:   {text}')
    print(f'Tokens: {tokens}')
    print(f'Words:  {decoded}')
    print()
```

## 2. Context window — how much an LLM can 'see'
The context window is the maximum number of tokens an LLM can process at once.

```python
# Context window sizes for popular models
models = {
    'GPT-2 (2019)':          1_024,
    'GPT-3 (2020)':          4_096,
    'GPT-4 (2023)':        128_000,
    'Claude 3.5 Sonnet':   200_000,
    'Gemini 1.5 Pro':    1_000_000,
    'Llama 3.1 (8B)':      128_000,
}

print(f'{'Model':<30} {'Context Window':>15} {'~Pages of Text':>16}')
print('-' * 64)
for model, ctx in models.items():
    pages = ctx // 500  # ~500 tokens per page
    bar = '█' * min(int(ctx / 50_000), 20)
    print(f'{model:<30} {ctx:>15,} {pages:>14,}p  {bar}')
```

## 3. Parameter count — what 'large' actually means

```python
import matplotlib.pyplot as plt

models = ['GPT-2\n(2019)', 'GPT-3\n(2020)', 'PaLM\n(2022)', 'GPT-4\n(est.)', 'Llama 3\n70B', 'Claude 3\nOpus (est.)']
params = [0.117, 175, 540, 1000, 70, 300]  # billions

fig, ax = plt.subplots(figsize=(12, 5))
fig.patch.set_facecolor('#0f1117')
ax.set_facecolor('#0f1117')
colors = ['#6c8ef5' if p < 200 else '#a78bfa' if p < 600 else '#f472b6' for p in params]
bars = ax.bar(models, params, color=colors, edgecolor='#2e3250', linewidth=0.8)
for bar, p in zip(bars, params):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 15,
            f'{p}B', ha='center', va='bottom', color='white', fontsize=10, fontweight='bold')
ax.set_ylabel('Parameters (billions)', color='white')
ax.set_title('LLM Parameter Count Over Time', color='white', fontweight='bold', pad=12)
ax.tick_params(colors='white')
ax.spines[:].set_color('#2e3250')
plt.tight_layout()
plt.show()
print('Note: GPT-4 and Claude sizes are estimates — companies do not publish exact counts.')
```
